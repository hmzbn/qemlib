```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import SparsePauliOp
from qemlib.rem import ReadoutMitigator, counts_executor, rem_expectation
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, depolarizing_error

noise_model = NoiseModel()
noise_model.add_all_qubit_quantum_error(depolarizing_error(0.02, 1), ["x", "h", "sx", "rz"])
backend = AerSimulator(noise_model=noise_model)

executor = counts_executor(backend, shots=4096)

mitigator = ReadoutMitigator(executor, n_qubits=2)
mitigator.calibrate()

qc = QuantumCircuit(2)
qc.h(0)

observable = SparsePauliOp.from_list([("XI", 1.0), ("ZZ", 0.5)])

value = rem_expectation(qc, observable, mitigator)
print("REM expectation:", value)
```
